```python
#D:
orders=[
{'category':'books'},
{'category':'clothes'},
{'category':'books'},
{'category':'toys'},
{'category':'books'},
{'category':'books'} 
]
category_count={}
for order in orders:
    category=order['category']
category_count[category]=category_count.get(category,0)+1
most_popular=max(category_count,key=category_count.get)
print('most popular category:',most_popular)
```
